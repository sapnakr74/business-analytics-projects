```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Q1: Load the dataset into the variable "sensors"
sensors = pd.read_csv("Sensors_M.csv")
print(sensors.head())
print(sensors.shape)

# Q2: Summary statistics and data errors
print(sensors.describe())
print(sensors.isnull().sum())          # any missing values?
print(sensors.dtypes)

# UnitsProduced has a negative min, which is impossible - let's look at those rows
print(sensors[sensors["UnitsProduced"] < 0])

# max UnitsProduced is 2100, far above the rest (~800-1500) - check it
print(sensors.sort_values("UnitsProduced", ascending=False).head(5))

# Also check that defective units are never more than units produced
print(sensors[sensors["DefectiveUnits"] > sensors["UnitsProduced"]])

# Errors found: 2 rows with negative UnitsProduced (Day 2 F02 and Day 23 F01)
# and 1 outlier row with 2100 units / 280 defects (Day 17 F02).
# I remove these 3 rows and continue with the cleaned data.
bad_rows = (sensors["UnitsProduced"] < 0) | (sensors["UnitsProduced"] > 1500)
sensors = sensors[~bad_rows].copy()
print(sensors.shape)

# Q3: Total output for each factory
factory_total = sensors.groupby("FactoryID")["UnitsProduced"].sum()
print(factory_total)

# Q4: Total output by factory and sensor type
factory_type_total = sensors.groupby(["FactoryID", "ProductLine"])["UnitsProduced"].sum()
print(factory_type_total)

# Q5: Which shift has the highest average output?
shift_avg = sensors.groupby("Shift")["UnitsProduced"].mean()
print(shift_avg)
print("Highest average output shift:", shift_avg.idxmax())

# Q6: Top 3 days with the highest total defect counts across all factories
day_defects = sensors.groupby("Day")["DefectiveUnits"].sum()
print(day_defects.sort_values(ascending=False).head(3))

# Q7: Add DefectRate = DefectiveUnits / UnitsProduced
sensors["DefectRate"] = sensors["DefectiveUnits"] / sensors["UnitsProduced"]
print(sensors.head())

# Q8: Combinations where defect rate exceeded 10%, and how frequent
high_defect = sensors[sensors["DefectRate"] > 0.10]
print(high_defect[["Day", "FactoryID", "Shift", "DefectRate"]])
print("Number of cases:", len(high_defect))
print("Share of all rows: {:.1%}".format(len(high_defect) / len(sensors)))
print(high_defect["FactoryID"].value_counts())   # which factories
print(high_defect["Shift"].value_counts())       # which shifts

# Q9a: Line plot of daily units produced per factory (all 3 on one graph)
daily = sensors.groupby(["Day", "FactoryID"])["UnitsProduced"].sum().reset_index()
plt.figure(figsize=(10, 5))
sns.lineplot(data=daily, x="Day", y="UnitsProduced", hue="FactoryID", marker="o")
plt.title("Daily Units Produced per Factory")
plt.show()

# Q9b: Bar chart of total units produced by shift and factory
shift_factory = sensors.groupby(["Shift", "FactoryID"])["UnitsProduced"].sum().reset_index()
plt.figure(figsize=(8, 5))
sns.barplot(data=shift_factory, x="Shift", y="UnitsProduced", hue="FactoryID")
plt.title("Total Units Produced by Shift and Factory")
plt.show()

# Q10: Two ways of calculating a factory's overall defect rate
avg_of_rates = sensors.groupby("FactoryID")["DefectRate"].mean()             # (a)
totals = sensors.groupby("FactoryID")[["DefectiveUnits", "UnitsProduced"]].sum()
ratio_of_totals = totals["DefectiveUnits"] / totals["UnitsProduced"]         # (b)
comparison = pd.DataFrame({"AvgOfRates": avg_of_rates, "RatioOfTotals": ratio_of_totals})
comparison["Difference"] = comparison["AvgOfRates"] - comparison["RatioOfTotals"]
print(comparison)
# They do NOT match exactly. (a) gives every row equal weight, so a small run
# counts as much as a big run. (b) weights each row by units produced, so
# it is the true share of defective units. (b) is more defensible.

# Q11: Synthesis - output, defect rate and downtime by factory and by shift
factory_summary = sensors.groupby("FactoryID").agg(
    TotalUnits=("UnitsProduced", "sum"),
    TotalDefects=("DefectiveUnits", "sum"),
    AvgDowntime=("DowntimeMinutes", "mean"),
    TotalDowntime=("DowntimeMinutes", "sum"))
factory_summary["DefectRate"] = factory_summary["TotalDefects"] / factory_summary["TotalUnits"]
print(factory_summary)

shift_summary = sensors.groupby("Shift").agg(
    AvgUnits=("UnitsProduced", "mean"),
    TotalDefects=("DefectiveUnits", "sum"),
    TotalUnits=("UnitsProduced", "sum"),
    AvgDowntime=("DowntimeMinutes", "mean"))
shift_summary["DefectRate"] = shift_summary["TotalDefects"] / shift_summary["TotalUnits"]
print(shift_summary)

# factory x shift to see the worst combination
combo = sensors.groupby(["FactoryID", "Shift"]).agg(
    AvgUnits=("UnitsProduced", "mean"),
    TotalDefects=("DefectiveUnits", "sum"),
    TotalUnits=("UnitsProduced", "sum"),
    AvgDowntime=("DowntimeMinutes", "mean"))
combo["DefectRate"] = combo["TotalDefects"] / combo["TotalUnits"]
print(combo)

# Q11 answer (based on the tables above):
# Factory: F03 is the one I would flag. It has the lowest total output (99,846),
# the highest defect rate (4.95%) and by far the highest average downtime
# (34.5 min vs about 29 for F01 and F02). It is the worst on all three metrics.
# Shift: Shift C is the one I would flag. It has the highest defect rate (5.04%),
# the highest average downtime (32.1 min) and below-B output. Shift B is the best on
# output and defect rate. Shift A has the lowest average output but lower downtime.
# Worst factory-shift combos: F01-C (5.2% defects) and F03-C (5.2% defects, 32 min downtime).
# Also, 4 of the 7 cases with defect rate above 10% happened in Shift C.
```
